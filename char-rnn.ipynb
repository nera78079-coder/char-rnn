```python
import tensorflow as tf
import numpy as np
```

```python
with open("data.txt") as f:
    content=f.read()
print(content)
```

```python
def preprocess(text):
    tokenizer = tf.keras.preprocessing.text.Tokenizer(char_level=True)
    tokenizer.fit_on_texts([text])

    max_id = len(tokenizer.word_index)

    encoded = np.array(
        tokenizer.texts_to_sequences([text])[0]
    ) - 1 

    return encoded, max_id,tokenizer 
```

```python
encoded, max_id, tokenizer= preprocess(content)

seq_length = 100 

x = []
y = []

for i in range(len(encoded) - seq_length):
    x.append(encoded[i:i+seq_length])
    y.append(encoded[i+1:i+seq_length+1])
x = np.array(x)
y = np.array(y)

print(x.shape)
print(y.shape)
```

```python
x = tf.one_hot(x, depth=max_id)
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.GRU(
        128,
        return_sequences=True,
        input_shape=[None, max_id],
        dropout=0.2,
        recurrent_dropout=0.2
    ),
    tf.keras.layers.GRU(
        128,
        return_sequences=True,
        dropout=0.2,
        recurrent_dropout=0.2
    ),
    tf.keras.layers.TimeDistributed(
        tf.keras.layers.Dense(max_id, activation="softmax")
    )
])
```

```python
model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam"
)
```

```python
model.fit(x,y,epochs=20)
```

```python
encoded, _,_ = preprocess("the fif")
x_new = tf.one_hot(encoded, depth=max_id) 
x_new = tf.expand_dims(x_new, axis=0)
y_pred = model.predict(x_new, verbose=0)
```

```python
next_id_class = tf.argmax(y_pred[0, -1]).numpy()

index_to_char = {
    idx - 1: char
    for char, idx in tokenizer.word_index.items()
}

print(index_to_char[next_id_class])
```
